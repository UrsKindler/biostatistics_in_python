### 1. Import Libraries


```python
from pathlib import Path
import sys
import importlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT_DIR = Path.cwd().parents[1]
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from Random_Proteomics_Dataset_Generator import load_template, create_random_proteomics_table, GROUP_TO_RUNS, COLOR_MAP

if '.' not in sys.path:
    sys.path.insert(0, '.')
mod = importlib.import_module('01b_pairwise_replicate_scatter')
make_pairwise_scatter_by_group = mod.make_pairwise_scatter_by_group
```

### 2. Load Proteomics Dataset


```python
try:
    df = load_template()
except Exception:
    df = create_random_proteomics_table(n_proteins=800, group_to_runs=GROUP_TO_RUNS, seed=42)
print(f'Proteins: {len(df)}, Columns: {df.shape[1]}')
df.head()
```

### 3. Generate Pairwise Scatter Plots


```python
out_path = make_pairwise_scatter_by_group(
    df_prepared=df,
    group_to_runs=GROUP_TO_RUNS,
    suffix='notebook_demo',
    outdir=Path('.'),
    color_map=COLOR_MAP,
    add_one_for_log2=True,
)
print(f'Saved: {out_path}')
```
